```python
# create an index list
import numpy as np
def create_array_idx(n, start):
    # Validate input
    if not isinstance(n, int) or n <= 0:
        raise ValueError("n must be a positive integer.")

    # Create array from 1 to n
    arr = np.arange(start, n + start)
    return arr
```
